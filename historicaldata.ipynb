# Get the historical data

```python
# pm_hourly_btc_midpoints.py
import os, csv, json, requests
from datetime import datetime, timedelta, timezone

# ---------------- CONFIG ----------------
N_DAYS = 2   # <-- change this for how many days back you want
OUTFILE = "btc_hourly_midpoints.csv"

GAMMA = "https://gamma-api.polymarket.com"
CLOB  = "https://clob.polymarket.com"
UTC = timezone.utc
```

## Helpers

```python
def get_events(start_dt: datetime, end_dt: datetime, limit=250, offset=0):
    params = {
        "limit": limit,
        "offset": offset,
        "order": "startDate",
        "ascending": True,
        "start_date_min": start_dt.isoformat().replace("+00:00","Z"),
        "start_date_max": end_dt.isoformat().replace("+00:00","Z"),
    }
    r = requests.get(f"{GAMMA}/events", params=params, timeout=30)
    r.raise_for_status()
    return r.json()

def normalize_array(raw):
    if raw is None: return []
    if isinstance(raw, str):
        s = raw.strip()
        if s.startswith("[") and s.endswith("]"):
            try:
                parsed = json.loads(s)
                if isinstance(parsed, list):
                    return [str(x) for x in parsed]
            except Exception: pass
        return [s]
    if isinstance(raw, (list, tuple)):
        return [str(x) for x in raw]
    return [str(raw)]

def map_up_down_indices(outcomes):
    lowered = [o.lower() for o in outcomes]
    up_idx = lowered.index("up") if "up" in lowered else 0 if outcomes else None
    down_idx = lowered.index("down") if "down" in lowered else (1 if len(outcomes) > 1 else None)
    if down_idx == up_idx and len(outcomes) >= 2:
        down_idx = 1 if up_idx == 0 else 0
    return up_idx, down_idx

def prices_history_minute(token_id: str, start_ts: int, end_ts: int):
    params = {"market": token_id, "startTs": start_ts, "endTs": end_ts, "fidelity": 1}
    r = requests.get(f"{CLOB}/prices-history", params=params, timeout=60)
    r.raise_for_status()
    payload = r.json()
    return payload.get("history", payload)
```

```python
def fetch_btc_hourly_midpoints(days=N_DAYS, outfile=OUTFILE):
    now = datetime.now(tz=UTC)
    start_window = now - timedelta(days=days)

    # get events in window (by event start)
    events = []
    offset = 0
    while True:
        batch = get_events(start_window, now, limit=250, offset=offset)
        if not batch:
            break
        events.extend(batch)
        if len(batch) < 250:
            break
        offset += 250

    # filter to btc up/down events only (event-level filter; market-level filter happens below)
    btc_events = [e for e in events if str(e.get("slug", "")).startswith("bitcoin-up-or-down-")]

    rows = []
    hour_secs = 3600
    tolerance = 120  # allow a little slack around exactly 3600s

    for ev in btc_events:
        slug = ev.get("slug")

        # fetch full event (to get all markets)
        ev_full = requests.get(f"{GAMMA}/events/slug/{slug}", timeout=30)
        if ev_full.status_code == 404:
            continue
        ev_data = ev_full.json()

        for market in (ev_data.get("markets") or []):
            if not market.get("enableOrderBook"):
                continue

            # parse market window
            try:
                sdt = datetime.fromisoformat(market.get("startDate").replace("Z", "+00:00"))
                edt = datetime.fromisoformat(market.get("endDate").replace("Z", "+00:00"))
            except Exception:
                continue

            # only 1-hour markets that started within window and have ended in the past
            dur = (edt - sdt).total_seconds()
            if not (abs(dur - hour_secs) <= tolerance):
                continue
            if not (start_window <= sdt <= now and edt <= now):
                continue

            # UP-only
            outcomes = normalize_array(market.get("shortOutcomes") or market.get("outcomes"))
            token_ids = normalize_array(market.get("clobTokenIds"))
            up_idx, _ = map_up_down_indices(outcomes)
            if up_idx is None or up_idx >= len(token_ids):
                continue

            tok = token_ids[up_idx]
            start_ts, end_ts = int(sdt.timestamp()), int(edt.timestamp())

            try:
                hist = prices_history_minute(tok, start_ts, end_ts)  # midpoint series
                for pt in hist:
                    ts = int(pt["t"])
                    iso_str = datetime.fromtimestamp(ts, tz=UTC).strftime("%Y-%m-%d %H:%M:%S")
                    rows.append([slug, "UP", tok, ts, iso_str, float(pt["p"])])
                print(f"{slug} UP (hourly) rows={len(hist)}")
            except Exception as e:
                print(f"{slug} UP error: {e}")

    # sort by time just to be tidy
    rows.sort(key=lambda r: (r[0], r[3]))

    # write single CSV
    with open(outfile, "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(["event_slug", "leg", "token_id", "timestamp", "iso_time_utc", "midpoint"])
        w.writerows(rows)

    print(f"\nWrote {len(rows)} rows to {outfile}")
```

```python
fetch_btc_hourly_midpoints()
```

# Plot the results

```python
import pandas as pd
import plotly.express as px

# load the CSV
df = pd.read_csv("btc_hourly_midpoints.csv")

# ensure datetime type
df["iso_time_utc"] = pd.to_datetime(df["iso_time_utc"])

# simple plot: one line across all rows
fig = px.line(
    df,
    x="iso_time_utc",
    y="midpoint",
    color="event_slug",     # separate lines per event
    hover_data=["token_id"],
    title="BTC UP Market Midpoints"
)

fig.update_layout(
    xaxis_title="UTC Time",
    yaxis_title="Midpoint",
    legend_title="Event"
)

fig.show()
```
